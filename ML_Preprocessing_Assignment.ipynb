# 🚀 ML Preprocessing Assignment  

## 📌 Objective  
**To apply categorical encoding and feature scaling techniques on a passenger dataset.**

# Dataset

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, RobustScaler

# Creating dataset
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
    'City': ['Mumbai', 'Delhi', 'Chennai', 'Mumbai', 'Delhi'],
    'Size': ['Small', 'Large', 'Medium', 'Large', 'Small'],
    'Age': [25, 45, 32, 60, 28],
    'Fare': [15, 300, 85, 450, 20]
}

df = pd.DataFrame(data)
df
```

## Task 1: Categorical Encoding  

- Gender → Label Encoding  
- City → One-Hot Encoding (drop first column)  
- Size → Ordered Encoding

```python
# Gender → Label Encoding]
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])

# City → One-Hot Encoding
df = pd.get_dummies(df, columns=['City'], drop_first=True, dtype = int)

# Size → Ordered Encoding
size_mapping = {'Small': 0, 'Medium': 1, 'Large': 2}
df['Size'] = df['Size'].map(size_mapping)

df
```

## Task 2: Feature Scaling  

**Applying RobustScaler due to presence of outlier in Fare column.**

```python
scaler = RobustScaler()

df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])

# RobustScaler is used because Fare contains an outlier (450),
# IUt scales data using median and IQR, making it robust to outliers.

df
```

## ✅ Conclusion  

**Successfully applied encoding and scaling techniques to prepare the dataset for machine learning.**
